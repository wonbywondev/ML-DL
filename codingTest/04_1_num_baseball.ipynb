```python
def solution(n, submit):

    count = 2

    head_S, head_B = int(submit(1234)[0]), int(submit(1234)[3])
    tail_S, tail_B = int(submit(5678)[0]), int(submit(5678)[3])

    head_count, tail_count = head_S + head_B, tail_S + tail_B


    if head_count == 4:
        exact_set = {str(i) for i in range(1, 5)}
        never_set = {str(i) for i in range(5, 10)}

    elif tail_count == 4:
        exact_set = {str(i) for i in range(5, 9)}
        never_set = {str(i) for i in range(1, 5)}

    else:
        exact_set = set()
        never_set = set()
    
        is_nine_included = head_count + tail_count == 4
        head_dict = {i: {str(n) for n in range(1, 5)} for i in range(4)}

        if is_nine_included:
            exact_set.add("9")
            tail_dict = {i: {str(n) for n in range(5, 10)} for i in range(4)}
        else:
            never_set.add("9")
            tail_dict = {i: {str(n) for n in range(5, 9)} for i in range(4)}


    ## 각 그룹에서 찾아내기
        for num in head_dict[0]:
            i = int(f"{num}000")
            count += 1 
            
            if int(submit(i)[0]) + int(submit(i)[3]) > 0:
                found_head += 1
                exact_set.add(num)
                if int(submit(i)[0]) > 0:
                    head_dict[0] = {str(num)}
                else:
                    head_dict[0] -= {str(num)}

                if found_head == head_count:
                    break
            else:
                never_set.add(num)
                for set_ in head_dict.values():
                    set_ -= str(num)


        for num in tail_dict[0]:
            i = int(f"{num}000")
            count += 1 

            if int(submit(i)[0]) + int(submit(i)[3]) > 0:
                found_tail += 1
                exact_set.add(num)
                if int(submit(i)[0]) > 0:
                    tail_dict[0] = {str(num)}
                else:
                    tail_dict[0] -= {str(num)}

                if found_tail == tail_count:
                    break
            else:
                never_set.add(num)
                for set_ in tail_dict.values():
                    set_ -= str(num)







    for i in range(1000, 10000):
        if count == 0:
            i = 1234

        if count == 1:
            i = 5678

        if submit(i) == "4S 0B": return i
    return 0
```
