```python
while True:
    input_ = input()
    LEN = len(input_)
    if 0 < LEN and LEN < 101 and input_.islower():
        break

if LEN == 1:
    print(1)

for i in range(1, LEN):
    if input_[i-1] != input_[-i]:
        print(0)
        break
    elif i == LEN-1:
        print(1)
```
