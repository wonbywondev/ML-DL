```python
import os

MODEL_NAME = "mnist-12-int8"
MODEL_URL = "https://github.com/onnx/models/raw/main/validated/vision/classification/mnist/model/mnist-12-int8.onnx"
MODEL_PATH = os.path.join("..", "models", MODEL_NAME, f"{MODEL_NAME}.onnx")
```

```python
import urllib.request

def download_model(url: str, path: str) -> None:
    """모델 파일이 없을 경우 다운로드"""
    if os.path.exists(path):
        print(f"이미 존재함: {path}")
        return

    os.makedirs(os.path.dirname(path), exist_ok=True)
    print(f"다운로드 중: {url}")
    urllib.request.urlretrieve(url, path)
    print(f"완료: {path}")

download_model(MODEL_URL, MODEL_PATH)
```
