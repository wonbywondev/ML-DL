```python
def solution(n, submit):
    candidate_dict = {
        0: [str(i) for i in range(1, 10)],
        1: [str(i) for i in range(0, 10)],
        2: [str(i) for i in range(0, 10)],
        3: [str(i) for i in range(0, 10)]
        }


    def submit(i):
        ANSWER = str(ANSWER)

        if ANSWER == i: 
            return "4S 0B"
        
        B = 0
        S = 0

        for idx, each_num in enumerate(i):
            if each_num == ANSWER[idx]:
                candidate_dict[idx] = each_num
                S += 1
            elif each_num in ANSWER:
                candidate_dict[idx].remove(each_num)
                B += 1
            else:
                for num_list in candidate_dict.values():
                    num_list.remove(each_num)

        return f"{S}S {B}B"

    count = 0
    while count < n:
        for i in range(1000, 10000):

            possible = True
            i = str(i)

            for idx, each_num in enumerate(i):
                if each_num not in candidate_dict[idx]:
                    possible = False
                    break

            if possible:
                count += 1
                if submit(i) == "4S 0B": 
                    return int(i)
        return 0
```
