```python
while True:
    try: 
        N = int(input())
        if 1 <= N and N <= 100:
            break
    except:
        print("1<=x<=100")

row_list = [[N-i, 2*i-1, N-i] for i in range(1, N+1)]
row_list.extend([row_list[-(i+2)] for i in range(len(row_list)-1)])

for cha_list in row_list:
    print(" "*cha_list[0]+"*"*cha_list[1])
```
