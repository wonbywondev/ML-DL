```python
def solution(n, submit):
    global count, prev_S, prev_B, prev_i

    prev_S = 0
    prev_B = 0
    
    count = 0

    semi_exact_set = set()
    never_set = set()
    
    candidate_dict = {
        0: {str(i) for i in range(1, 10)},
        1: {str(i) for i in range(0, 10)},
        2: {str(i) for i in range(0, 10)},
        3: {str(i) for i in range(0, 10)}
        }
    
    def this_is_B(num: str, index: int):
        pass

    def this_is_S(num: str, index: int):
        pass

    def this_is_X(num: str, index: int):
        pass
    
    def consider(i):
        possible = True
        str_i = str(i)

        if semi_exact_set - set(str_i):
            possible = False
            return possible
        
        if set(str_i) - never_set != set(str_i):
            possible = False
            return possible

        for idx, each_num in enumerate(str_i):
            if each_num not in candidate_dict[idx]:
                possible = False
                break

        return possible


    for i in range(1000, 10000):
        if count >= n:
            break

        if not consider(i):
            continue
        
        count += 1

        if submit(i) == "4S 0B": 
            return i
        
        else:
            now_i = str(i)
            now_S = int(submit(i)[0])
            now_B = int(submit(i)[3])

            if now_S + now_B == 0:
                for set_ in candidate_dict.values():
                    set_ = set_ - set(now_i)

            elif now_S + now_B == 4:
                semi_exact_set.update(now_i)
                for idx in range(0,4):
                    candidate_dict[idx] = set(candidate_dict[idx]) & set([num for num in now_i])
                    
            if count == 1:
                prev_i = now_i
                prev_S = now_S
                prev_B = now_B
                continue

            same_index = [idx for idx in range(0,4) if now_i[idx] == prev_i[idx]]
            diff_index = [idx for idx in range(0,4) if now_i[idx] != prev_i[idx]]
            diff_count = len(diff_index)

            diff_S = now_S - prev_S
            diff_B = now_B - prev_B

            is_S_diff = now_S != prev_S
            is_B_diff = now_B != prev_B

            if diff_count == 1:
                if is_S_diff:
                    if diff_S > 0:
                        candidate_dict[diff_index[0]] = now_i[diff_index[0]]
                        semi_exact_set.add(now_i[diff_index[0]])
                    else:
                        candidate_dict[diff_index[0]] = candidate_dict[diff_index[0]] - {now_i[diff_index[0]]}
                        semi_exact_set.add(prev_i[diff_index[0]])

                elif is_B_diff:
                    if diff_B > 0:
                        candidate_dict[diff_index[0]] = candidate_dict[diff_index[0]] - {now_i[diff_index[0]]}
                        semi_exact_set.add(now_i[diff_index[0]])
                    else:
                        candidate_dict[diff_index[0]] = candidate_dict[diff_index[0]] - {prev_i[diff_index[0]]}
                        semi_exact_set.add(prev_i[diff_index[0]])

            elif diff_count == 2:
                if is_S_diff and not is_B_diff:
                    if diff_S == 2:
                        pass
                    elif diff_S == 1:
                        pass
                    elif diff_S == -1:
                        pass
                    elif diff_S == -2:
                        pass
                    pass
                elif is_S_diff and is_B_diff:
                    pass
                elif not is_S_diff and is_B_diff:
                    pass

            prev_i = now_i
            prev_S = now_S
            prev_B = now_B


    return 0
```
