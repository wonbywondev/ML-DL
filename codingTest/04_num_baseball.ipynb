```python
def solution(n, submit):
    global count, prev_S, prev_B, prev_i
   
    exact_set = set()
    never_set = set()
    
    candidate_dict = {
        0: {str(i) for i in range(1, 10)},
        1: {str(i) for i in range(1, 10)},
        2: {str(i) for i in range(1, 10)},
        3: {str(i) for i in range(1, 10)}
        }

    
    def consider(i):
        str_i = str(i)

        if exact_set - set(str_i):
            return False
        
        if set(str_i) & never_set:
            return False
        
        if len(set(str_i)) != 4:
            return False

        for idx, each_num in enumerate(str_i):
            if each_num not in candidate_dict[idx]:
                return False

        return True



    if int(submit(1234)[0]) + int(submit(1234)[3]) + int(submit(5678)[0]) + int(submit(5678)[3]) == 4:
        never_set.add("9")
        for value_ in candidate_dict.values():
            value_.remove("9")
    else:
        exact_set.add("9")

    count = 2
    prev_i = "1234"
    prev_S = int(submit(1234)[0]) 
    prev_B = int(submit(1234)[3])

    for i in range(1000, 10000):
        if not consider(i):
            continue

        if submit(i) == "4S 0B": 
            return i
        elif count == n:
            break
        
        now_i = str(i)
        now_S = int(submit(i)[0])
        now_B = int(submit(i)[3])

        if now_S + now_B == 0:
            never_set.update(now_i)
            for idx in range(4):
                candidate_dict[idx] -= set(now_i)

        elif now_S + now_B == 4:
            exact_set.update(now_i)
            for idx in range(4):
                candidate_dict[idx] &= set(now_i)

        diff_index = [idx for idx in range(4) if now_i[idx] != prev_i[idx]]
        diff_count = len(diff_index)

        if diff_count == 1:
            idx = diff_index[0]
            if now_S != prev_S:
                if now_S > prev_S:
                    candidate_dict[idx] = {now_i[idx]}
                    exact_set.add(now_i[idx])
                    for i in range(4):
                        if i != idx:
                            candidate_dict[i] -= {now_i[idx]}

                else:
                    candidate_dict[idx] = {prev_i[idx]}
                    exact_set.add(prev_i[idx])
                    for i in range(4):
                        if i != idx:
                            candidate_dict[i] -= {prev_i[idx]}

            if now_B != prev_B:
                if now_B > prev_B:
                    candidate_dict[idx] -= {now_i[idx]}
                    exact_set.add(now_i[idx])
                else:
                    candidate_dict[idx] -= {prev_i[idx]}
                    exact_set.add(prev_i[idx])


        prev_i = now_i
        prev_S = now_S
        prev_B = now_B

    return 0
```
